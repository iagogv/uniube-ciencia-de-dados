```python
import numpy as np

# Arrays 1D
a = np.array([1, 2, 3, 4, 5])
print('1D:', a, 'shape:', a.shape)

# Arrays 2D
b = np.zeros((3, 4))
c = np.arange(12).reshape(3, 4)
print('2D:\n', c)
```

## Fonte 1 - CSV público (UCI)

```python
import pandas as pd

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'
wine = pd.read_csv(url, sep=';')
print('Wine:', wine.shape)
wine.head()
```

## Fonte 2 - seaborn

```python
import seaborn as sns

tips = sns.load_dataset('tips')
print('Tips:', tips.shape)
tips.head()
```

## Fonte 3 - OpenML

```python
from sklearn.datasets import fetch_openml

iris = fetch_openml('iris', version=1, as_frame=True, parser='auto')
iris_df = iris.frame
print('Iris OpenML:', iris_df.shape)
iris_df.head()
```

## Fonte 4 - JSON público (semi-estruturado)

```python
url_json = 'https://raw.githubusercontent.com/vega/vega-datasets/master/data/cars.json'
cars = pd.read_json(url_json)
print('Cars JSON:', cars.shape)
cars.head()
```
